```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df= pd.read_csv("athlete_events.csv")
df.head()
```

```python
df[df["Year"].isin([2016])].count()
```

```python
summer_2016 = df[df["Year"] == 2016]
summer_2016
```

```python
NOC_meadal_count = summer_2016.groupby(["NOC", "Sex"])["Medal"].count().unstack(fill_value= 0)

fig, ax = plt.subplots(figsize = (40, 15))
ax.plot(NOC_meadal_count.index, NOC_meadal_count["M"], label = "M", color = "k")
ax.plot(NOC_meadal_count.index, NOC_meadal_count["F"], label = "F", color = "red")
plt.ylabel("Medals")
plt.xlabel("NOC names")
plt.title("Medals")
plt.legend(title = "SEX")
plt.xticks(rotation = 90)
plt.show()
```

```python
gold = summer_2016[summer_2016["Medal"] == "Gold"]
NOC_medals_gold_sex = gold.groupby(["NOC", "Sex"])["Medal"].count().unstack(fill_value= 0)
fig, ax = plt.subplots(figsize = (20, 8))
ax.plot(NOC_medals_gold_sex.index, NOC_medals_gold_sex["M"], color = "k", label = "Male and medals of gold")
ax.plot(NOC_medals_gold_sex.index, NOC_medals_gold_sex["F"], color = "red", label = "Female and medals of gold")
plt.legend()
plt.xlabel("NOC names")
plt.ylabel("Count medal of gold")
plt.title("Summer 2016 olpic")
plt.xticks(rotation = 90)
plt.show()
```

```python
gold = summer_2016[summer_2016["Medal"] == "Gold"]
NOC_medals_gold_sex = gold.groupby(["NOC", "Sex"])["Medal"].count().unstack(fill_value= 0)

plt.style.use("ggplot")
fig, ax = plt.subplots(figsize = (20, 8))
ax.plot(NOC_medals_gold_sex.index, NOC_medals_gold_sex["M"], label = "Male medal gold", color = "k")
ax.plot(NOC_medals_gold_sex.index, NOC_medals_gold_sex["F"], label = "Female medal gold", color = "red")
plt.xticks(rotation = 90)
plt.xlabel("NOC name")
plt.ylabel("Number of gold medals")
plt.title("Summer 2016 olipc")
plt.legend(title = "Sex")
plt.show()
```

```python
gold = summer_2016[summer_2016["Medal"] == "Gold"]
NOC_medals_gold_sex = gold.groupby(["NOC", "Sex"])["Medal"].count().unstack(fill_value= 0)

fig , ax = plt.subplots(figsize = (20, 8))

ax.bar(NOC_medals_gold_sex.index, NOC_medals_gold_sex["M"], label = "Male medal gold", color = "k", alpha = 0.4)
ax.bar(NOC_medals_gold_sex.index, NOC_medals_gold_sex["F"], label = "Female medal gold", color = "red", alpha = 0.4)
plt.title("Summer 2016")
plt.xlabel("NOC names")
plt.ylabel("Number of gold medals")
plt.xticks(rotation = 90)
plt.legend(title = ("Sex"))
plt.show()
```

```python
noc_medal_size = summer_2016.groupby("NOC")["Medal"].count().reset_index()
noc_medal_size.rename(columns={"Medal": "Medal Count"}, inplace=True)

sns.relplot(
    data= noc_medal_size,
    x="NOC",
    y="Medal Count",
    kind="scatter",
    size="Medal Count",
    height= 8,
    aspect= 2
)
plt.xticks(rotation = 90)
plt.show()
```

inplace=True → DataFrame'ni joyida o‘zgartiradi (overwrite).
inplace=False → Hech narsani o‘zgartirmaydi, faqat yangi DataFrame qaytaradi.

reset_index() — bu indeksni oddiy ustunga aylantiradi va yangi standart raqamli indeks beradi.

```python
noc_medal_size = summer_2016.groupby(["NOC"])["Medal"].count().reset_index()
noc_medal_size.rename(columns= {"Medal" : "Medal_Count"}, inplace= True)

sns.relplot(
            data= noc_medal_size,
            x = "NOC",
            y = "Medal_Count",
            kind = "scatter",
            size = "Medal_Count",
            hue= "Medal_Count",
            height= 8,
            aspect= 3
            )
plt.xticks(rotation = 90)
plt.show()
```
