```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

```python
df= pd.read_csv("car_prices.csv")
df.head()
```

```python
plt.figure(figsize = (12, 6))
plt.hist(df["brand"], bins = 30, edgecolor = "black" )
plt.ylabel("Number")
plt.xlabel("Car name")
plt.title("number of cars")
plt.xticks(rotation = 90)
plt.legend("CAR")
plt.show()
```

```python
df.head()
```

```python
df1 = pd.read_csv("summer2016.csv")
df1.head()
```

```python
brand_price = df.pivot_table(values= "price", index= "brand", columns= "fuel_type" , aggfunc= np.mean)
brand_price.plot(kind = "bar", stacked= True, figsize= (16, 6))
plt.ylabel("Car price")
plt.xlabel("Brand car")
plt.title("Car and price")
plt.grid(axis = "y" )
plt.legend(title = "Fuel")
plt.xticks(rotation = 90)
plt.show()
```

```python
df.head()
```

```python
brand_milage = df.pivot_table(
    values="milage",
    index="brand",
    columns="fuel_type",
    aggfunc=np.sum
)

brand_milage.loc["Ford"].plot(
    kind="pie",
    figsize=(10, 10),
    autopct="%1.1f%%",
    title="Toyota Milage by FuelType"
)
plt.ylabel("")
plt.show()
```

```python
brand_price1 = df.pivot_table(
    values= "price",
    index= "brand",
    aggfunc= np.mean
)

brand_price1.plot(
    kind = "pie",
    subplots= True,
    figsize= (25, 25),
    autopct="%1.1f%%",
    title= "Car price"
)

plt.show()
```

```python
top5 = df.groupby("brand")["price"].max().sort_values(ascending= False).head(5)
top5.plot(kind = "pie", subplots= True, figsize= (25, 25), autopct="%1.1f%%")
plt.legend(title = "Max price car")
plt.show()
```

```python
df.head()
```

```python
brand_model_year = df.groupby("brand")["model_year"].count().sort_values().head(10)
brand_model_year.plot(
    kind = "pie",
    subplots= True,
    figsize= (10, 10),
    autopct = "%1.1f%%",
    title= "Brand modey year"
)

plt.show()
```

```python
brand_model_year.plot(
    kind = "bar",
    stacked= True,
    figsize= (10, 5),
    color = "c"
)
plt.xlabel("Brand cars")
plt.ylabel("Years")
plt.title("Brands")
plt.show()
```

```python
plt.figure(figsize = (10, 5))
plt.hist(df1["Age"], bins = 20, edgecolor = "black")
plt.xlabel("Age")
plt.ylabel("Nums")
plt.show()
```

```python
fig, ax = plt.subplots(3, 2, figsize = (12, 10))
ax[0, 0].hist(df1["Age"], bins = 50, edgecolor = "black")
ax[0, 0].set_title("Age Histogram", size = 10)
ax[0, 1].hist(df1["Height"], bins = 50 , edgecolor = "black", color = "y")
ax[0, 1].set_title("Height Histogram", size = 10)
ax[1, 0].hist(df1["Weight"], bins = 50, edgecolor = "black", color = "r")
ax[1, 0].set_title("Weight Histogram", size = 10)
ax[1, 1].plot(df1["Age"].unique(), marker = "o")
ax[1, 1].set_title("Age Plot", size = 10)
ax[2, 0].plot(df1["Height"].unique(), marker = "o")
ax[2, 0].set_title("Height Plot", size = 10)
ax[2, 1].plot(df1["Weight"].unique(), marker = "o")
ax[2, 1].set_title("Weight Plot")
plt.show()
```

```python
M_F = df1.pivot_table(
    columns= "Medal",
    index= "Sex",
    aggfunc= "size",
    fill_value= 0
)
M_F.plot(kind = "bar", stacked= False, figsize= (8, 4), color = ["gold", "silver", "#cd7f32"] )
plt.xlabel("Sex")
plt.ylabel("Medal")
plt.xticks(rotation = 0)
plt.title("SEX")
plt.show()
```

```python
df1.head()
```

```python
Name_NOC = df1.groupby(["NOC", "Medal"]).size().unstack(fill_value= 0)
Name_NOC.plot(kind = "bar", subplots= False, figsize= (20, 15), edgecolor = "black", color = ["brown", "gold", "silver"])
plt.xlabel("NOC")
plt.ylabel("Counts")
plt.xticks(rotation = 90)
plt.show()
```

```python
Name_NOC = df1.groupby(["NOC", "Medal"]).size().unstack(fill_value= 0)
Name_NOC.plot(kind = "bar", subplots= True, figsize= (20, 10), edgecolor = "black", color = ["brown", "gold", "silver"])
plt.xlabel("NOC")
plt.ylabel("Counts")
plt.xticks(rotation = 90)
plt.show()
```
