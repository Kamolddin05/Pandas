```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
fig, ax = plt.subplots() # rasm + grafik chizadigan joyini yasaydi. 
plt.show()
# fig ---> butun rasm
# ax ---> rasm ichida bitta grafik oynasi
```

```python
df = pd.read_csv("car_prices.csv")
df.head()
```

```python
brand_price_grb = df.groupby("brand")["price"].mean()
fig, ax = plt.subplots(figsize=(12, 8))

ax.plot(brand_price_grb.index, brand_price_grb.values, color = "red")
ax.set_xlabel("Brand")
ax.set_ylabel("Average Price")
ax.set_title("Average Price by Brand")
plt.xticks(rotation=90) 
plt.show()
```

```python
fig, ax = plt.subplots(figsize=(12, 8))
ax.plot(brand_price_grb.index, brand_price_grb.values, marker ="v")
ax.set_ylabel("Average Price")
ax.set_xlabel("Brand")
ax.set_title("Average Price by Brand")
plt.xticks(rotation=90)
plt.show()
```

```python
fig, ax = plt.subplots(figsize=(12, 8))
ax.plot(brand_price_grb.index, brand_price_grb.values, marker ="o")
ax.set_ylabel("Average Price")
ax.set_xlabel("Brand")
ax.set_title("Average Price by Brand")
plt.xticks(rotation=90)
plt.show()
```

```python
fig, ax = plt.subplots(figsize = (12, 6))
ax.plot(brand_price_grb.index, brand_price_grb.values, marker = "v", linestyle = "--", color = "red")
ax.set_xlabel("Car name")
ax.set_ylabel("Car price mean")
ax.set_title("Car")
plt.xticks(rotation = 90)
plt.show()
```
