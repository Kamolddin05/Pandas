```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv("summer2016.csv")
df.head()
```

```python
medals = df.pivot_table(index= "NOC", columns= "Medal", values= "Sex", aggfunc= "size", fill_value= 0)
fix, ax = plt.subplots(figsize = (20, 10))
ax.bar(medals.index, medals["Gold"], color = "gold")
plt.title("Count gold and NOC")
plt.ylabel("Count medals")
plt.xlabel("NOC names")
plt.xticks(rotation = 90)
plt.legend("Medals")
plt.show()

```

```python
medals1 = df.pivot_table(index = "NOC", columns= "Medal", values = "Sex", aggfunc= "size", fill_value= 0)
fig, ax = plt.subplots(figsize = (20, 9))
ax.plot(medals1.index, medals1["Gold"], color = "gold", marker = "o")
ax.plot(medals1.index, medals1["Bronze"], color = "brown", marker = "o")
ax.plot(medals1.index, medals1["Silver"], color = "silver", marker = "o")
ax.set_ylabel("Numbers of medals")
ax.set_xlabel("NOC names")
plt.xticks(rotation = 90)
plt.show()
```

```python
df.head()
```

```python
fig, ax = plt.subplots(figsize = (20, 8))
ax.bar(medals.index, medals["Gold"], color = "gold", label = "Gold")
ax.bar(medals.index, medals["Silver"], color = "silver", bottom= medals["Gold"], label = "Silver")
ax.bar(medals.index, medals["Bronze"], color = "brown", bottom= medals["Gold"] + medals["Silver"], label = "Bronze")
ax.set_title("Medals statistiks of NOC")
ax.set_xlabel("NOC names")
ax.set_ylabel("Numbers of medals")
plt.xticks(rotation = 90)
plt.legend(title = "Medals of taypes")
plt.show()
```

```python
sports = df.pivot_table(index= "Sport", values= ["Height", "Weight"], aggfunc= "mean" , fill_value= 0)
filter_sport = sports.loc[["Rowing", "Gymnastics"]]
fig, ax = plt.subplots(figsize = (20, 9))
ax.bar(filter_sport.index, filter_sport["Height"], color = "k")
ax.bar(filter_sport.index, filter_sport["Weight"], color = "red")
ax.set_xlabel("Sports of names")
ax.set_ylabel("Nums")
ax.set_title("Height and Weight")
plt.xticks(rotation = 0)
plt.show()
```

```python
rowing_data = df.loc[df["Sport"].isin(["Rowing"]), "Height"]
gymnastics_data = df.loc[df["Sport"].isin(["Gymnastics"]), "Height"]
fig, ax = plt.subplots(figsize=(20, 9))
ax.hist(rowing_data, color="b", label = "Rowing", histtype= "step")
ax.hist(gymnastics_data, color="red", label = "Gymnastics", histtype= "step")
plt.xlabel("Height (cm)")
plt.ylabel("Frequency")
plt.title("Rowing Height Distribution")
plt.legend()
plt.show()
```

```python
df.head()
```

```python
sport_rowing = df[df["Sport"] == "Rowing"]
sport_gymnastics = df[df["Sport"] == "Gymnastics"]
fix, ax = plt.subplots(figsize = (12, 6))
ax.bar(["Rowing"], sport_rowing["Height"].mean(), yerr = sport_rowing["Height"].std(), color = "b")
ax.bar(["Gymnastics"], sport_gymnastics["Height"].mean(), yerr = sport_rowing["Height"].std(), color = "red")
ax.set_xlabel("Names of sports")
ax.set_ylabel("Height")
ax.set_title("Height mean")
plt.show()
```

```python
medals_err = df.pivot_table(index= "NOC", columns= "Medal", aggfunc= "size", fill_value= 0)
fig, ax = plt.subplots(figsize = (20, 9))
ax.errorbar(medals_err.index, medals_err["Gold"], color = "gold", marker = "o", label = "Gold")
ax.errorbar(medals_err.index, medals_err["Silver"], color = "silver", marker = "o", label = "Silver")
ax.errorbar(medals_err.index, medals_err["Bronze"], color = "brown", marker = "o", label = "Bronze")
ax.set_xlabel("NOC")
plt.xticks(rotation = 90)
ax.set_ylabel("Nums of medals")
plt.legend(title = ("Medals of tayps"))
plt.show()

```
