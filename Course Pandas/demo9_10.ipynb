```python
import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv("athlete_events.csv")
df.head()
```

```python
medals_sex_gold = df[df["Medal"] == "Gold"].pivot_table(index = "NOC", columns= "Sex", aggfunc= "size", fill_value= 0)

fig, ax = plt.subplots(figsize = (25, 6))

ax.plot(medals_sex_gold.index, medals_sex_gold["M"], color = "k", label = " M Gold")
ax.plot(medals_sex_gold.index, medals_sex_gold["F"], color = "red", label = " F Gold")
ax.set_ylabel("Counts of medal gold")
ax.set_xlabel("Noc names")
plt.legend(title = ("Medal GOLD"))
plt.xticks(rotation = 90)
plt.show()
```

```python
medals = df["Medal"].value_counts()

fig, ax = plt.subplots(figsize = (12, 6))
ax.bar(medals.index, medals.values)
ax.set_xticklabels(medals.index, rotation = 0)
ax.set_xlabel("Medal type")
ax.set_ylabel("Number of medals")
ax.set_title("Medal distribution")
plt.show()
```

```python
sports = df["Sport"].unique()
fig, ax = plt.subplots(figsize = (25, 10))

for sport in sports:
    sport_df = df[df["Sport"] == sport]
    ax.bar(sport, sport_df["Height"].mean(),
           yerr = sport_df["Height"].std())
ax.set_ylabel("Height (cm)")
ax.set_xticklabels(sports, rotation = 90)
plt.show()
```

```python
SexM_count = df.groupby("NOC")["Sport"].count().reset_index() # reset_index() — DataFrame indeksini oddiy ustun qilib qayta tiklaydigan pandas metodi.

sns.relplot(x="NOC", 
            y="Sport", 
            data=SexM_count,
            color="blue", 
            kind ="line",
            height= 12,
            aspect= 2,
            alpha = 1,
            )

plt.xticks(rotation=90)
plt.ylabel("Number of male athletes")
plt.show()
```
