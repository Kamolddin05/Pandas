```python
import pandas as pd
df = pd.read_csv("seattle-weather (1).csv")
df.head()
```

```python
import pandas as pd
import matplotlib.pyplot as plt

df['date'] = pd.to_datetime(df['date'])

df['year_month'] = df['date'].dt.to_period('M')

monthly_stats = df.groupby('year_month')['temp_max'].agg(
    Monthly_Min='min',
    Monthly_Max='max',
    Monthly_Mean='mean'
).reset_index()


plt.plot(monthly_stats['year_month'].astype(str), monthly_stats['Monthly_Min'], label='Min Temperature')
plt.plot(monthly_stats['year_month'].astype(str), monthly_stats['Monthly_Max'], label='Max Temperature')
plt.plot(monthly_stats['year_month'].astype(str), monthly_stats['Monthly_Mean'], label='Mean Temperature')

plt.xlabel("Year-Month")
plt.ylabel("Temperature")
plt.title("Monthly Temperature Statistics")
plt.xticks(rotation=90)
plt.legend()
plt.tight_layout()
plt.show()



```
