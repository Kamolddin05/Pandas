```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv("summer2016.csv")
df.head()
```

```python
plt.figure(figsize= (10, 6))
plt.hist(df["Age"], bins = 20, edgecolor = "black")
plt.xlabel("Age")
plt.ylabel("Count Hummens")
plt.title("Age distribution")
plt.grid(True, color = "r")
plt.show()
```

```python
M_F_count_medal = df.pivot_table(index= "Sex", columns= "Medal", aggfunc= "size", fill_value= 0)
M_F_count_medal.plot(kind = "bar", stacked=False, figsize=(15, 6), color=["gold", "silver", "#cd7f32"])
plt.title("Jins boyicha taqsimot")
plt.xlabel("Sex")
plt.ylabel("Medal count")
plt.xticks(rotation = 0)
plt.legend(title = "Medal")
plt.grid(axis = "y")
plt.show()
```
