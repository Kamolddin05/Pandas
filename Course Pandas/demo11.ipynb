```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv("athlete_events.csv")
df.head()
```

```python
summer_2012 = df[df["Year"].isin([2012])]
summer_2012.head()
```

```python
sns.countplot(data=summer_2012[summer_2012.notna()],
              x = "Medal")
plt.show()
```

```python
sns.catplot(data = summer_2012[summer_2012.notna()],
            x = "Medal",
            kind = "count")
plt.show()
```

```python
category_order = ["Silver", "Gold", "Bronze"]
sns.catplot(data = summer_2012[summer_2012.notna()],
            x = "Medal",
            kind = "count",
            order= category_order)
plt.show()
```
